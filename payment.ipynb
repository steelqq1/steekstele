```python
import base64
from IPython.display import Image, display

mermaid_code = """flowchart TD
    A([Начало]) --> B[/"Ввести баланс s и цену p"/]
    B --> C{"s ≥ p?"}
    C -->|Да| D[/"Вывести «Оплата прошла»"/]
    C -->|Нет| E[/"Вывести «Недостаточно средств»"/]
    D --> F([Конец])
    E --> F
"""

encoded = base64.urlsafe_b64encode(mermaid_code.encode("utf-8")).decode("ascii")
live_url = "https://mermaid.live/edit#base64:" + encoded
svg_url = "https://mermaid.ink/svg/" + encoded

print("Редактор:", live_url)
print("Картинка:", svg_url)
display(Image(url=svg_url))
```

```python
def result(s, p):
    return "Оплата прошла" if s >= p else "Недостаточно средств"

tests = [
    (100, 100, "Оплата прошла"),
    (99, 100, "Недостаточно средств"),
    (500, 100, "Оплата прошла"),
    (0, 100, "Недостаточно средств"),
]

for s, p, expected in tests:
    got = result(s, p)
    assert got == expected, f"s={s}, p={p}: ожидалось {expected}, получено {got}"
    print(f"s={s}, p={p} -> {got}")

print("Все проверки пройдены")
```
